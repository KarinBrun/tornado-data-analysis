#### Import and dataframe setup

```python
# importing library
import pandas as pd

# reading in dataframes
allDF = pd.read_csv('1950-2025_all_tornadoes.csv')
print(allDF.head(2))
actualDF = pd.read_csv('1950-2025_actual_tornadoes.csv')
print(actualDF.head(2))
```

```python
# NOAA has two different databases of tornado information. One is all_tornadoes.csv which is a raw data dump that includes all state and continuing county segments, and actual_tornadoes.csv which tracks tornadoes individually without state and county segments (https://www.spc.noaa.gov/wcm/#data)
```

#### Data Exploration

```python
# data exploration of all_tornadoes.csv
print(f'Base shape of All dataframe: {allDF.shape}\n')
print(f'Dataframe columns: {allDF.columns.tolist()}\n')
print('Dataframe data types:')
print(f'{allDF.dtypes}\n')

print('Dataframe Info:')
print(f'{allDF.info()}\n')
print('Null values:')
print(allDF.isnull().sum())
```

```python
# data exploration of actual_tornadoes.csv
print(f'Base shape of Actual dataframe: {actualDF.shape}\n')
print(f'Dataframe columns: {actualDF.columns.tolist()}\n')
print('Dataframe data types:')
print(f'{actualDF.dtypes}\n')

print('Dataframe Info:')
print(f'{actualDF.info()}\n')
print('Null values:')
print(actualDF.isnull().sum())
```

```python
# I decided to use the actual_tornadoes.csv file instead of the all_tornadoes.csv file since I am not using the stf, stn, ns, sn, or sg columns. (Read https://www.spc.noaa.gov/wcm/data/SPC_severe_database_description.pdf for details on columns)
```

#### Data Cleaning

```python
# dropping/keeping columns
actualDF_kept = actualDF[['om', 'yr', 'mo', 'dy', 'date', 'time', 'st', 'mag', 'inj', 'fat', 'loss', 'slat', 'slon', 'elat', 'elon', 'len', 'wid']]

display(actualDF_kept.head())
print(f'Shape of dataframe with only kept columns: {actualDF_kept.shape}')

# keeping columns om (number associated with specific tornado for categorizing purposes), yr (year), mo (month), dy (day), date (exact date), time (exact time tornado began), st (state), mag (F/EF rating), inj (number of injuries), fat (number of fatalities), loss (amount of damage in monetary value), slat (starting latitude), slon (starting longitude), elat (ending latitude), elon (ending longitude), len (length of track in miles), wid (width in yards)

# dropping columns tz (timezone not necessary), stf (state fips not needed), stn (state number not needed), closs (crop loss only began tracking after 2007, not enough data), ns, sn, sg (all three are unnecessarily complicated for project), f1, f2, f3, f4 (all four are state fips codes, not being used for project), fc (adding complication to database that isn't needed for project)
```

```python
# checking dataframe after cleaning
print(f'Dataframe columns: {actualDF_kept.columns.tolist()}\n')
print('Dataframe data types:')
print(f'{actualDF_kept.dtypes}\n')

print('Dataframe Info:')
print(f'{actualDF_kept.info()}\n')
print('Null values:')
print(actualDF_kept.isnull().sum())
```

```python
# no datatypes need to be changed in code, changing columns to categories will be done in Tableau with calculated fields to make data readable
```

```python
# sorting by date
actualDF_sorted = actualDF_kept.sort_values(by='date', ignore_index=True)
print(actualDF_sorted.head())
print(actualDF_sorted.shape)
```

#### Exporting Dataframe

```python
# creating new file
actualDF_sorted.to_csv('us_tornado_dataset_1950_2025.csv', index=False)
```
